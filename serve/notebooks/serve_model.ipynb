# Serve the fine-tuned model

Loads the pushed model, exposes it over HTTP via ngrok, and (optionally) tells the deployed API where to find it. Works in Colab, Kaggle, or RunPod (or any GPU box) - see `serve/notebooks/*.py`'s own "Cell N" comments, which this notebook's cells below mirror directly.

**Secrets/env vars** - Colab/Kaggle: set these via the key icon in the left sidebar (**Secrets**), and grant this notebook access when prompted. RunPod or any plain GPU box (no notebook secrets vault): set the same names as real environment variables before starting Jupyter (`export NAME=value`, or your pod's env-var config) - `get_secret()` in every script below falls back to `os.environ` automatically outside Colab/Kaggle, no code change needed.

| Name | Required | Used by |
|---|---|---|
| `HF_USER` | yes | to build the model repo name to load |
| `HF_TOKEN` | only if that model repo is private | read access is enough here |
| `NGROK_AUTH_TOKEN` | yes | from [ngrok.com](https://ngrok.com) - needed to open the tunnel |
| `ENDPOINT_AUTH_TOKEN` | yes | bearer token this endpoint requires from callers - set it to whatever `financial-sentiment-api` is configured to send |
| `MODEL_CHOICE` | optional | override the base model family to load |
| `MODEL_VERSION` | optional | override the pushed version to load |
| `RENDER_SECRET_KEY` | only for the "update backend" cell | must match the API_KEY the deployed API checks |

Run cells 1-2 in order; skip cell 3 ("update backend") if you're not pointing a live deployment at this session. Cell 4 ("keep alive") is Colab-specific (a no-op elsewhere) and optional either way.

```python
import os
import subprocess

if "_REPO_BASE" not in globals():
    _REPO_BASE = os.getcwd()
_REPO_PATH = os.path.join(_REPO_BASE, "financial-sentiment-model")

if os.path.isdir(_REPO_PATH):
    subprocess.run(["git", "pull", "--ff-only"], cwd=_REPO_PATH, check=True)
else:
    subprocess.run(["git", "clone", "--depth", "1", "https://github.com/GiulianoAparecido/financial-sentiment-model.git", _REPO_PATH],
                   check=True)

os.chdir(_REPO_PATH)
```
