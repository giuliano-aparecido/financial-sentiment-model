# Evaluate the base model only

Standalone fallback for when you just need the BASE-model comparison on its own (e.g. you already have `evaluate_model.py`'s fine-tuned numbers from an earlier run and don't want to redo that pass). Works in Colab or Kaggle (RunPod doesn't offer TPUs).

**Secrets/env vars** - Colab/Kaggle: set these via the key icon in the left sidebar (**Secrets**), and grant this notebook access when prompted. Outside Colab/Kaggle (no notebook secrets vault, and no TPU runtime on RunPod anyway): set the same names as real environment variables instead - `get_secret()` in every script below falls back to `os.environ` automatically, no code change needed.

| Name | Required | Notes |
|---|---|---|
| `HF_USER` | yes | to build the target repo name |
| `HF_TOKEN` | only if your HF model repo is private | read access is enough here - this script never pushes |
| `MODEL_CHOICE` | optional | must match whatever the target push was actually trained under |
| `MODEL_VERSION` | optional | which push to reload |

**Needs** the `dataset_val*.jsonl` files already present in the working directory - either from running the train_and_evaluate notebook earlier in this same session (the setup cell below reuses that same clone via `git pull`, it does NOT delete it), or uploaded manually into `financial-sentiment-model/` after running the setup cell.

Run all cells in order (Runtime -> Run all).

```python
import os
import subprocess

if "_REPO_BASE" not in globals():
    _REPO_BASE = os.getcwd()
_REPO_PATH = os.path.join(_REPO_BASE, "financial-sentiment-model")

if os.path.isdir(_REPO_PATH):
    subprocess.run(["git", "pull", "--ff-only"], cwd=_REPO_PATH, check=True)
else:
    subprocess.run(["git", "clone", "--depth", "1", "https://github.com/GiulianoAparecido/financial-sentiment-model.git", _REPO_PATH],
                   check=True)

os.chdir(_REPO_PATH)
```
